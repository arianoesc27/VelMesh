```python
import matplotlib.pyplot as plt
import numpy as np

def plot_model(coords, ax, color='k'):
    """Función para trazar el modelo con los puntos dados."""
    edges = [(0, 1), (1, 2), (2, 3), (3, 0)]  # Conexiones entre los puntos
   
    ax.set_xticks(range(-3, 4))
    ax.set_yticks(range(-3, 4))
    ax.axhline(0, color='gray', linewidth=1)  # Eje X
    ax.axvline(0, color='gray', linewidth=1)  # Eje Y
    ax.set_xlim(-3, 3)
    ax.set_ylim(-3, 3)
    ax.set_aspect('equal')
    ax.grid(True, linestyle='--', alpha=0.5)

    for edge in edges:
        p1, p2 = coords[edge[0]], coords[edge[1]]
        ax.plot([p1[0], p2[0]], [p1[1], p2[1]], color+'-', linewidth=2)
    
    # Etiquetar los puntos
    for i, (x, y) in enumerate(coords):
        ax.text(x, y, f'C{i+1}', fontsize=12, verticalalignment='bottom', horizontalalignment='right', color=color)

# Coordenadas originales con C1 como origen
coords_original = [(0, 0), (1, 0), (1, -2), (0, -2)]
coords_rotated = [(0, 0), (0, 1), (2, 1), (2, 0)]

# Graficar ambas figuras en una sola
fig, ax = plt.subplots(figsize=(6, 6))
plot_model(coords_original, ax,'k')  # Modelo original en negro
plot_model(coords_rotated, ax, 'r')   # Modelo rotado en verde

# Agregar punto rojo en C1
ax.plot(0, 0, 'ro', markersize=8)

ax.set_title('Original (black) and Rotated -90° (red)')
plt.savefig('modelrot', dpi=300, bbox_inches='tight')
plt.show()
```
